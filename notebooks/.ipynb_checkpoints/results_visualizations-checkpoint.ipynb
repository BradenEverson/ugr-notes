# Result Visualizations from first Semester

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
starvations = pd.read_csv("../data/starvation_times.csv")
starvations.head()
```

```python
rr = starvations[starvations["scheduler"] == "rr"]["average starvation (ms)"]
rr_dt = starvations[starvations["scheduler"] == "rr-dt"]["average starvation (ms)"]
rr_svm = starvations[starvations["scheduler"] == "rr-svm"]["average starvation (ms)"]

plot_labels = ["Round Robin", "Round Robin - Decision Tree Quantum", "Round Robin - SVM Quantum"]

plt.figure(figsize=(10, 6)) 
plt.boxplot([rr, rr_dt, rr_svm], labels=plot_labels)
plt.title("Starvation (STM ticks) by the Different Schedulers")
plt.xlabel("Scheduler")
plt.ylabel("Average Starvation per Task (STM ticks)")
```
