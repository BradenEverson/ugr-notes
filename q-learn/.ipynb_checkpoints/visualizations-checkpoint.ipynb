## Part 1. Simple Reward Function

To begin, we'll create the simplest reward function possible: 

### $$R = t_{cpu} - t_{wait}$$

Where 

 - $t_{cpu}$: Percentage of time spent between last context switch and now running on the CPU
 - $t_{wait}$: Percentage of time spent between last context switch and now waiting (not including IO)


We will test this model versus a baseline of always allocating 10ms time quantums in a variety of different process configurations

### Scenario 1: 1 IO Heavy Process and 1 CPU Heavy Process

```python
plt.plot(baseline["avg_starvation"], label="Q-Learning Agent")
plt.plot(q_reward["avg_starvation"], label="Baseline")

plt.title("Baseline Simulation vs. Q-Learning Agent With Simple Reward")

plt.legend()

plt.ylabel("Average Starvation %")
plt.xlabel("Timestamp")
```

### Scenario 2: Many CPU Heavy Processes with Variable Runtimes before small IO Interrupts

```python
q_reward = pd.read_csv("../data/sim/simple_3_cpu_heavy.csv")
baseline = pd.read_csv("../data/sim/baseline_3_cpu_heavy.csv")
```

```python
plt.plot(q_reward["avg_starvation"], label="Q-Learning Agent")
plt.plot(baseline["avg_starvation"], label="Baseline")

plt.title("Baseline Simulation vs. Q-Learning Agent With Simple Reward")

plt.legend()

plt.ylabel("Average Starvation %")
plt.xlabel("Timestamp")
```
